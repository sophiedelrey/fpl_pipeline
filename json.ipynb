```python
# the data
import requests
import pandas as pd

base_url = 'https://fantasy.premierleague.com/api/'
response = requests.get(base_url + 'bootstrap-static/')
summary = response.json()

# the main structure
print("Main sections in the API response:")
print(summary.keys())

# Explore ELEMENTS (players)
print("\n=== ELEMENTS (Players) ===")
print(f"Total players: {len(summary['elements'])}")
print("\nFirst player example:")
import json
print(json.dumps(summary['elements'][0], indent=2))

# Explore TEAMS
print("\n=== TEAMS ===")
print(f"Total teams: {len(summary['teams'])}")
print("\nFirst team example:")
print(json.dumps(summary['teams'][0], indent=2))

# Explore ELEMENT_TYPES (Positions)
print("\n=== ELEMENT TYPES (Positions) ===")
print(json.dumps(summary['element_types'], indent=2))

# Explore EVENTS (Gameweeks)
print("\n=== EVENTS (Gameweeks) ===")
print(f"Total gameweeks: {len(summary['events'])}")
print("\nFirst gameweek example:")
print(json.dumps(summary['events'][0], indent=2))
```
